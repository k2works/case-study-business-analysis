# ファイルから内容をテキストに抽出するJupyter Notebook
このノートブックでは、PDFやWordファイルなどの内容をテキスト形式に抽出する方法を示します。

## 必要なライブラリのインポート
Pythonの標準ライブラリや、PyPDF2やdocxなどの外部ライブラリをインポートします。

```python
# 必要なライブラリのインポート
import os
from docx import Document
import pdfplumber
```

## テキスト抽出対象ファイルの読み込み
抽出対象のファイル（例: PDFやWordファイル）を指定して読み込みます。

```python
# テキスト抽出対象ファイルの読み込み
file_path = "2019年度事例4.pdf"  # 抽出対象のファイルパスを指定
if not os.path.exists(file_path):
    raise FileNotFoundError(f"指定されたファイルが見つかりません: {file_path}")
```

## テキスト抽出処理
ファイル形式に応じたライブラリを使用して、内容をテキスト形式に抽出します。

```python
# テキスト抽出処理
def extract_text_from_pdf(file_path):
    text = ""
    with pdfplumber.open(file_path) as pdf:
        for page in pdf.pages:
            text += page.extract_text() or ""
    return text

def extract_text_from_docx(file_path):
    doc = Document(file_path)
    text = "\n".join([paragraph.text for paragraph in doc.paragraphs])
    return text

# ファイル形式に応じた処理
if file_path.endswith(".pdf"):
    extracted_text = extract_text_from_pdf(file_path)
elif file_path.endswith(".docx"):
    extracted_text = extract_text_from_docx(file_path)
else:
    raise ValueError("対応していないファイル形式です。PDFまたはWordファイルを指定してください。")
```

## 抽出結果の保存
抽出したテキストをファイルに保存するか、画面に出力します。

```python
# 抽出結果の保存
output_file = "extracted_text.txt"
with open(output_file, "w", encoding="utf-8") as f:
    f.write(extracted_text)

print(f"抽出したテキストを {output_file} に保存しました。")
```
